# 01 — Data Exploration

Explore the structure and statistics of the kernel scheduling DAG data.

```python
import sys
sys.path.insert(0, '../src')

from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from ks_core.graph import load_json, list_cases

DATA_DIR = Path('../data/raw/json')
sns.set_theme(style='whitegrid')
```

```python
# List all available cases
cases = list_cases(DATA_DIR, fmt='json')
print(f'Available cases: {cases}')
```

```python
# Load and summarize each case
summaries = []
for case in cases:
    inst = load_json(DATA_DIR / f'{case}.json')
    ops = [n for n in inst.nodes if n.op not in ('ALLOC', 'FREE')]
    allocs = [n for n in inst.nodes if n.op == 'ALLOC']
    summaries.append({
        'case': case,
        'total_nodes': len(inst.nodes),
        'total_edges': len(inst.edges),
        'compute_ops': len(ops),
        'alloc_ops': len(allocs),
        'total_buf_size': sum(n.size for n in allocs),
    })

df = pd.DataFrame(summaries)
df
```

```python
# Operation type distribution for each case
for case in cases:
    inst = load_json(DATA_DIR / f'{case}.json')
    op_counts = pd.Series([n.op for n in inst.nodes]).value_counts()
    print(f'\n=== {case} ===')
    print(op_counts)
```
